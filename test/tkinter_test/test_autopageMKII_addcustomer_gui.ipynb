```python
from tkinter import *
```

```python
window1 = Tk()
window1.title("Autosamatic")
```

---

# CUSTOM WINDOWS

> Window1

```python
window1.geometry("800x600")
window1.configure(bg="#444")
```

---

# FUNCTIONS

```python
def search():
    search_query = entered_order.get()
    print(search_query)
    # report_log.config(text=f"นั่นล่ะOrder: {search_query}")
```

---

# **WIDGETS**

> Labels

```python
label_order = Label(window1, text="Order: ")
label_order.pack()
```

> Inputs

```python
entered_order = StringVar()
order_input = Entry(window1, textvariable=entered_order)

order_input.pack()
```

> Buttons

```python
search_btn = Button(window1, text="ค้นหา", bg="#747474", command=search)
search_btn.pack()
```

> Log windows

```python
report_log = Text(window1, state=DISABLED)
report_log.pack(side='bottom', fill=X, padx=10, pady=(0,10))
```

---

# **RUN GUI**


```python
window1.mainloop()
```

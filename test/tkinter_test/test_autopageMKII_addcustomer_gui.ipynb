```python
from tkinter import *
```

```python
window1 = Tk()
window1.title("Autosamatic")
```

---

# CUSTOM WINDOWS

> Window1

```python
window1.geometry("800x600")
window1.configure(bg="#444")
```

---

# **WIDGETS**

> Labels

```python
label_order = Label(window1, text="Order: ")
label_order.pack()
```

> Inputs

```python
order_input = Entry(window1)
order_input.pack()
```

> Buttons

```python
search_btn = Button(window1, text="ค้นหา", bg="#747474")
search_btn.pack()
```

> Log windows

```python
report_log = Text(window1, state=DISABLED)
report_log.pack(side='bottom', fill=X, padx=10, pady=(0,10))
```

---

# **RUN GUI**


```python
window1.mainloop()
```

## >Pattern1 เกือบจะดี

```python
import tkinter as tk
from tkinter import ttk

class ScrollableFrame(tk.Frame):
    def __init__(self, parent, *args, **kwargs):
        super().__init__(parent, *args, **kwargs)
        parent.title("Serial Filla!!")
        # parent.geometry("800x600")
        
        #* สร้าง Canvas ทำหน้าที่เป็นหน้าจอหลัก หากมีการปรับขนาด Gui ทั้งหมด จะมีขนาด หดลง หรือใหญ่ขึ้น และทำหน้าที่เป็นพื้นหลัง นิ่งๆ
        self.canvas = tk.Canvas(self, borderwidth=0)
        self.canvas.config(bg="pink")
        #* สร้าง widget scrollbar //orient="vertical": คือการแสดงหน้าตา widget เป็นแบบแนวตั้ง และนอกจากนี้ 'vertical' เป็นค่า default ด(คือไม่setก็ได้ ถ้าจะใช้แนวตั้ง)// self.canvas.yview ถ้าไม่ใช้ propนี้ เราจะ interact ด้วยการคลิกลากไม่ได้
        self.scrollbar = ttk.Scrollbar(self, orient="vertical", command=self.canvas.yview)
        #* สร้าง Frame ที่ไว้ใช้สำหรับ scroll แล้วเอาไปผูกเป็นลูกของ canvas และ Frame นี้จะเป็นส่วนที่มีขนาด Dynamic ยืดหดได้ จะถูกนำไปใช้เป็น BG ให้ method create_window
        self.scrollable_frame = tk.Frame(self.canvas)
        
        
        #* bind คือการเชื่อม event เข้ากับ method ซึ่ง event ในที่นี่คือ configure: ซึ่งมันคือ eventที่เมื่อ Frame ถูกปรับขนาด (Configure) ซึ่งขนาดที่เพิ่มขึ้นจะถูกกำหนดอัตโนมัติเมื่อ Frame มี widgetภายในที่มากขึ้น !!!แต่!!! ขอบเขตscaleของ Scrollbar นั้นไม่เปลี่ยนแปลงตาม-
        #* -จึงต้องมี Event ที่ดักไอ้การเปลี่ยนแปลงที่เกิดแบบอัตโนมัตินี้ เมื่อมีการเปลี่ยนแปลงของพื้นที่(Configure)ในwidgetที่เลือกไว้(ตัวอย่างคือ Frame)
        self.scrollable_frame.bind("<Configure>", self._configure_scroll_region)

        #* เขาเลือกที่จะใช้ Create_window กับส่วนที่เป็น scrollable 
        self.canvas.create_window((0, 0), window=self.scrollable_frame, anchor="nw")
        self.canvas.configure(yscrollcommand=self.scrollbar.set)

        self.canvas.pack(side="left", fill="both", expand=True)
        self.scrollbar.pack(side="right", fill="y")

        self.canvas.bind_all("<MouseWheel>", self._on_mousewheel)

    #* Method สำหรับกำหนดขอบเขตการ Scroll โดยคืนค่าพิกัด ของพื้นที่ทั้งหมดของ canvas แล้วใช้เป็น % สำหรับการ Scroll
    def _configure_scroll_region(self, event):
        #* ที่มันต้องใช้กับ canvas เพราะว่ามัน เราจะดูเพื้นที่ที่ถูกใช้ ใน canvas ซึ่งใน tkinter มันมี widget หลักๆสองอันคือ canvas กับ frame,  เราจะดูว่า frame(ตัวเล็ก) กินพื้นที่ canvas(ตัวใหญ่) ไปเท่าไหร่ 
        self.canvas.configure(scrollregion=self.canvas.bbox("all"))

    def _on_mousewheel(self, event):
        #* การเลื่อน ลูกกลิ้งขึ้น มันจะ event ส่งค่า delta เป็นลบ แต่ scrollbar เมื่อได้รับค่าลบ มันจะเลื่อนลง เลยต้องใช้ -1* เพื่อให้ค่ากลับมาบวกเพื่อให้ scroll แสดงผลว่าเลื่อนขึ้น scroll ที่อยู๋บนมีค่ามากสุดคือ 1 และล่างสุดคือ 0
        self.canvas.yview_scroll(int(-1 * (event.delta / 120)), "units")

# Example usage:
root = tk.Tk()
scrollable_frame = ScrollableFrame(root)
# for i in range(50):
#     tk.Label(scrollable_frame.scrollable_frame, text=f"Label {i}").pack()
tk.Label(scrollable_frame.scrollable_frame, text=f"Label").pack()
scrollable_frame.pack(fill="both", expand=True)
root.mainloop()
```

## >Pattern3

```python
class MainApp(tk.Frame):
    def __init__(self, parent, *args, **kwargs):
        super().__init__(parent, *args, **kwargs)
        parent.title("Auto Serial!")
        parent.geometry("400x300")

        # Canvas
        self.canvas = tk.Canvas(self, borderwidth=0)
        self.scrollbar = ttk.Scrollbar(self, orient='vertical', command=self.canvas.yview)
        self.scrollable_frame = tk.Frame(self.canvas)

        self.scrollable_frame.bind("<Configure>", self._on_frame_configure)

        self.canvas.create_window((0, 0), window=self.scrollable_frame, anchor="nw")
        self.canvas.configure(yscrollcommand=self.scrollbar.set)

        self.canvas.pack(side="left", fill="both", expand=True)
        self.scrollbar.pack(side="right", fill="y")

        self.canvas.bind_all("<MouseWheel>", self._on_mousewheel)

    def _on_frame_configure(self, event):
        self.canvas.configure(scrollregion=self.canvas.bbox("all"))

    def _on_mousewheel(self, event):
        self.canvas.yview_scroll(int(-1 * (event.delta / 120)), "units")
        
# Example usage:
root = tk.Tk()
scrollable_frame = ScrollableFrame(root)
for i in range(10):
    tk.Label(scrollable_frame.scrollable_frame, text=f"Label {i}").pack()
tk.Label(scrollable_frame.scrollable_frame, text=f"Label").pack()
scrollable_frame.pack(fill="both", expand=True)
root.mainloop()
```

### > get data from import file

```python
import pandas as pd
import numpy as np
import math

class UnifyData:
    def __init__(self, path):
        self.dir = path
        self.excel_file = pd.ExcelFile(self.dir)
        self.sheet_names = self.excel_file.sheet_names
          
        # * อยากให้col ไหนเป็น int มาเติมที่ list 'must_be_int_cols'
        must_be_int_cols = ['qty']
        
        self.data_state = {}
        self.column_names = []
        for sheet_name in self.sheet_names:
            self.df = pd.read_excel(self.dir, sheet_name=sheet_name, dtype=str)

            # * ปรับ dtype ของ column ตาม list 'must_be_int_cols' ที่กำหนดไว้ 
            if any(must_be_int_cols) in self.df.columns:
                for col in must_be_int_cols:
                    try:
                        self.df[col].astype(int)
                    except:
                        continue

            self.lowercase_columns = [column.lower() for column in self.df.columns]
            self.column_names = [column for column in self.df.columns]
            self.df.columns = self.lowercase_columns
            if 'spec' in self.lowercase_columns:
                print(f"ค่าของ Sheet {sheet_name} จะถูกจัดเก็บใน key items_list")
                self.data_state['items_list'] = self.df
            elif 'set' in self.lowercase_columns:
                print(f"Sheet: {sheet_name} จะถูกจัดเก็บใน key sn")
                self.data_state['sn'] = self.df

        # print(f"items_list: {self.data_state['items_list']}")
        # print(f"items_list type: {type(self.data_state['items_list'])}")
        # print(f"sn: {self.data_state['sn']}")
        # print(f"sn type: {type(self.data_state['sn'])}")
        # print(f"self.data_state: {self.data_state}")
        # print(f"self.data_state: {pd.DataFrame(self.data_state)['sn']}")
    def get_result(self, sku, set_number):
        #* คัดเอา data sku ที่ต้องการ
        find_dict = self.data_state['sn'][self.data_state['sn']['set'] == set_number].iloc[0]
        
        #* ตัด nan
        result = {}
        for key, value in find_dict.items():
            if not isinstance(value, float) or not np.isnan(value):
                result[key] = value
        
        #* รวม me1
        me1_values = [value for key, value in result.items() if key.startswith("me1_")]
        result["me1"] = " ".join(me1_values)
        
        #* Return Result
        if sku == result['code']:
            print('all')
            print(result)
            return result
        else:
            print('Code และ เลขชุดset ไม่ถูกต้อง')
            print(f"Code {sku}")
            print(f"เลขชุดSet {set_number}")
        
file_path = r"C:\Users\CSH0041\Downloads\DATA Program nHack V3.xlsx"
sku_input = 'KCU2-000795'
set = '795/001'

unified_data = UnifyData(file_path)
unified_data.get_result(sku_input, set)
```

```python
import threading, time
from tkinter import * 
from customtkinter import *



#* Functions for test
def shorter_function():
    print('Starting shorter')
    for i in range(3):
        time.sleep(1)
        i += 1
        print(i)
        textbox.insert(END, f"{i}\n")

    print('Exiting shorter')
    
def longer_function():
    print('Starting longer')
    for i in range(10):
        time.sleep(1)
        i += 1
        print(i)
        textbox.insert(END, f"{i}\n")

    print('Exiting longer')

shorter_thread = threading.Thread(target=shorter_function)
longer_thread = threading.Thread(target=longer_function)

#* test
# print(f"is shorter_thread alive? : {shorter_thread.is_alive()}")
# shorter_thread.start()

# print(f"is shorter_thread alive? : {shorter_thread.is_alive()}")
# shorter_thread.join()


##* Gui3 Scrollbar ง่ายจ้าดดดด
app = CTk()
app.title("Autosmatic")
app.geometry("300x200")



#* functions
def check_threads(shorter_thread_cycle, longer_thread_cycle, callback=None):
    #* หาก Thread ใด Thread หนึ่ง ทำงานอยู่ ให้เช็คตัวเองอีกรอบ ภายในเวลา 100 millisec 
    if shorter_thread_cycle.is_alive() or longer_thread_cycle.is_alive():
        #* after(เวลาmillisec, callbackfunction)
        app.after(100, lambda: check_threads(shorter_thread_cycle, longer_thread_cycle, callback))
    else:
        #* เมื่อ Thread ทั้งสองไม่ alive แล้วให้ทำงาน รวม thread เข้ากับ thread หลัก แล้วเรียกใช้ callback ถ้าหากมี callback มาด้วยน่ะนะ callbackนี้จะรับ operation_startเข้ามาให้ทำงานอีกรอบ
        shorter_thread_cycle.join()
        longer_thread_cycle.join()
        print("shorter_thread_cycle is alive?: ", shorter_thread_cycle.is_alive())
        print("longer_thread_cycle is alive?: ", longer_thread_cycle.is_alive())
        if callback:
            callback()
            

def operation_start(self, callback=None):
    #* สร้าง Thread 
    shorter_thread_cycle = threading.Thread(target=shorter_function)
    longer_thread_cycle = threading.Thread(target=longer_function)
    #* สั่ง Thread ให้เริ่มทำงาน
    shorter_thread_cycle.start()
    longer_thread_cycle.start()
    #* ตรวจสอบว่า Thread ทั้งสองยังทำงานอยู่หรือไม่
    check_threads(shorter_thread_cycle, longer_thread_cycle, callback)
    
def accel_mode_start(self):
    n = 3
    def start_next_cycle(count):
        if count < n:
            textbox.insert(END, f"ทำงานรอบที่ {count + 1}\n")
            operation_start(lambda: start_next_cycle(count + 1))
        else:
            self.display_bot_status_label.config(text=f"Bot Status: ᕦʕ •ᴥ•ʔᕤ กำลังทำงาน", bg="#cf1313", fg="#ffffff")
            pass
        
    textbox.insert(END, "ทำงานรอบที่ 1\n")
    operation_start(lambda: start_next_cycle(1))
    
    
#* สร้าง CTkScrollableFrame เพื่อใส่ Widgets
scrollable_frame = CTkScrollableFrame(master=app)
scrollable_frame.pack(expand=True, fill=BOTH)

#* เพิ่ม Widgets ลงใน CTkScrollableFrame
start_btn = CTkButton(master=scrollable_frame, text="Start!!", command=operation_start)
start_btn.pack()

accel_start_btn = CTkButton(master=scrollable_frame, text="Accel_Start!!", command=accel_mode_start)
accel_start_btn.pack()



#* textbox
textbox = CTkTextbox(scrollable_frame)
textbox.pack()



# for i in range(50):
#     text = StringVar(value=f"Label {i}")  # กำหนดค่าเริ่มต้นให้กับ StringVar
#     entry = CTkEntry(master=scrollable_frame, textvariable=text, state='normal')
#     entry.insert(0, text.get())  # แทรกค่าเริ่มต้นลงใน Entry
#     entry.configure(state='readonly')
#     entry.pack()


app.mainloop()



# #* Gui1
# app = CTk() 

# # def on_mousewheel(event):
# #     canvas.yview_scroll(-1* (event.delta // 120), "units")
# app.geometry("150x200")
# set_default_color_theme("DaynNight.json")

# frame = CTkScrollableFrame(master=app, fg_color="#8D6F3A", border_color="#FFCC70" ,border_width=2, orientation="vertical", scrollbar_button_color=("#DB3E39", "#4AAD11"))
# frame.pack(expand=True, fill=BOTH)

# # canvas = Canvas(app)
# # canvas.pack(side=LEFT, fill=BOTH, expand=True)

# # scroll_bar = Scrollbar(app, orient=VERTICAL, command=canvas.yview)
# # scroll_bar.pack(side=RIGHT, fill=Y)

# # canvas.configure(yscrollcommand=scroll_bar.set)
# # canvas.bind('<Configure>', lambda e: canvas.configure(scrollregion=canvas.bbox("all")))
# # canvas.bind_all('<MouseWheel>', on_mousewheel)

# # frame = Frame(canvas)
# # canvas.create_window((0,0), window=frame, anchor="nw")



# for line in range(1, 8):
#     button = CTkButton(master=frame, text=str(line)).pack(expand=True)
    
# app.mainloop()

# #* Gui2
# app = CTk()
# app.grid_rowconfigure(0, weight=1)
# app.grid_columnconfigure(0, weight=1)

# # create scrollable textbox
# tk_textbox = CTkTextbox(app, activate_scrollbars=False)
# tk_textbox.grid(row=0, column=0, sticky="nsew")

# # create CTk scrollbar
# ctk_textbox_scrollbar = CTkScrollbar(app, command=tk_textbox.yview)
# ctk_textbox_scrollbar.grid(row=0, column=1, sticky="ns")

# # connect textbox scroll event to CTk scrollbar
# tk_textbox.configure(yscrollcommand=ctk_textbox_scrollbar.set)

# app.mainloop()




```

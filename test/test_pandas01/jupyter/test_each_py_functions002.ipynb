# Test ย้าย ค่าใน Table

```python
import pandas as pd

# สร้าง DataFrame ตัวอย่าง
data = {'col1': ['ค่า1', 'ค่า2', 'ค่า3'],
        'col2': ['ข้อมูล1', 'ข้อมูล2', 'ข้อมูล3'],
        'col3': ['เงื่อนไข', 'เงื่อนไข', 'เงื่อนไข']}
df = pd.DataFrame(data)
print(df)

# ดำเนินการตามเงื่อนไข โดยย้ายค่าจาก col1 ไป col3 เมื่อค่า1ถูกนำไปดำเนินการบางอย่างแล้ว
df.loc[df['col1'] == 'ค่า1', 'col3'] = df['col1']
df.loc[df['col1'] == 'ค่า1', 'col1'] = ''

# แสดง DataFrame หลังจากดำเนินการ
print(df)

# เขียน DataFrame ลงในไฟล์ Excel และเขียนทับไฟล์ Excel เดิม
df.to_excel('ชื่อไฟล์.xlsx', index=False)
```

```python
import re

order_note = r"""บริษัท เอเวอร์ ท็อป อินเตอร์เนชั่นแนล จำกัด (สำนักงานใหญ่)
Tax ID:0105539059785
เลขที่ 1406 ถนน มาเจริญ แขวง หนองแขม เขต หนองแขม กรุงเทพมหานคร 10160
"""

def note_extractor():
        if order_note != 'nan':
            try:
                name_match = re.search(r'ชื่อ:(.*?)\n', order_note)
                branch_match = re.search(r'สาขา:(.*?)\n', order_note)
                address_match = re.search(
                    r'ที่อยู่:(.*?)\n', order_note)
                tax_id_match = re.search(r'Tax id:(.*?)', order_note)
                print("regexบันทึก ชื่อ: ", name_match)
                print("ใช้ group กับ regexบันทึก: ", name_match.group(1))
            except:
                print("Exception")
                name_match = re.search(r'บริษัท.*', order_note)
                branch_match = re.search(r'สาขา:(.*?)\n', order_note)
                address_match = re.search(
                    r'ที่อยู่:(.*?)\n', order_note)
                tax_id_match = re.search(r'\d{13}', order_note)
                print("regexบันทึก บริษัท: ", name_match)
                print("except name_match: ", name_match.group()) if name_match else print("name_matchไม่มีค่า")
                print("except branch_match: ", branch_match.group()) if branch_match else print("branch_match ไม่มีค่า")

                print("except address_match: ", address_match.group()) if address_match else print("address_match ไม่มีค่า")
                print("except tax_id_match: ", tax_id_match.group()) if tax_id_match else print("tax_id_match ไม่มีค่า")
        else:
            print("ไม่มีค่า")
            
note_extractor()        
```

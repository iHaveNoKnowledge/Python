```python
# สร้าง DataFrame ตัวอย่าง
data = {'ชื่อลูกค้า': ['John', 'Alice', 'Bob', 'David'],
        'ที่อยู่ลูกค้า': ['123 ถนน A', '456 ถนน B', '789 ถนน C', '101 ถนน D'],
        'รายการสินค้า': ['สินค้า A', 'สินค้า B', 'สินค้า A', 'สินค้า C']}
data_frame = pd.DataFrame(data)

# เลือกเฉพาะคอลัมน์ที่คุณต้องการ
selected_columns = data_frame[['ชื่อลูกค้า', 'ที่อยู่ลูกค้า', 'รายการสินค้า']]

# แสดงข้อมูลที่เลือก
print("ข้อมูลทั้งหมด:")
print(selected_columns)

# เข้าถึงค่าแบบแถวแรกของคอลัมน์ 'ชื่อลูกค้า'
first_customer_name = selected_columns['ชื่อลูกค้า'].iloc[0]
print("\nชื่อลูกค้าแรก:", first_customer_name)

# เข้าถึงค่าแบบแถวทั้งหมดของคอลัมน์ 'รายการสินค้า'
product_list = selected_columns['รายการสินค้า']
print("\nรายการสินค้าทั้งหมด:")
print(product_list)

# เข้าถึงค่าแบบแถวแรกของคอลัมน์ 'ที่อยู่ลูกค้า' แล้วแปลงเป็น List
first_address_list = selected_columns['ที่อยู่ลูกค้า'].iloc[0].split()
print("\nที่อยู่ลูกค้าแรก (แบ่งเป็น List):", first_address_list)

# เข้าถึงค่าแบบแถวแรกของคอลัมน์ 'รายการสินค้า' แล้วแปลงเป็นตัวแปรปกติ (ไม่ใช่ Pandas Series)
first_product = selected_columns['รายการสินค้า'].iloc[0]
print("\nรายการสินค้าแรก (เป็นตัวแปรปกติ):", first_product)
```

# Base64 decoder

```python
import base64

data = {"filename":"B0183-E01-0000002-payment.pdf","filetype":"application/pdf","base64":""}


pdf_data = base64.b64decode(data['base64'])

with open(data['filename'], 'wb') as f:
    f.write(pdf_data)
print(f"บันทึกไฟล์แล้ว: {data['filename']}")
```

# Base64 management


```python
import pdfplumber
import os
import base64

base64_001_001 = ""

def pdf_to_txt(file_path):
    result = ""
    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            result += page.extract_text()
    return result


pdf_path = f"output.pdf"


def base64_to_pdf(base64_string):
    pdf_bytes = base64.b64decode(base64_string)
    with open(f"{pdf_path}", "wb") as pdf_file:
        pdf_file.write(pdf_bytes)

def print_pdf(pdf_path):
    # os.startfile(pdf_path, "print")
    print("execute!! ")
    print("os.startfile(pdf_path, 'print')")

inv_poped_up = "B0183-E01-0000001"

#* create pdf
base64_to_pdf(base64_001_001)

#* collect txt from pdf
extracted_txt = pdf_to_txt(pdf_path)
print(f"""extracted_txt:\n{extracted_txt}\n""")

#* is inv correct?
if inv_poped_up in extracted_txt:
    print(f"""inv_poped_up:\ncorrect inv!!\n{inv_poped_up}""")
    #* print
    print_pdf(pdf_path)
else:
    print(f"""inv_poped_up:\nwrong inv!!\nget src again""")
    
```

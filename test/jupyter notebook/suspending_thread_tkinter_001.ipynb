```python
import tkinter as tk
from threading import Thread
from queue import Queue, Empty
import time

class App:
    def __init__(self, root):
        self.root = root
        self.queue = Queue()
        self.running = False
        
        self.start_button = tk.Button(root, text="Start", command=self.start)
        self.start_button.pack()
        
        self.stop_button = tk.Button(root, text="Stop", command=self.stop)
        self.stop_button.pack()

        # Label for displaying messages
        self.label = tk.Label(root, text="Ready")
        self.label.pack()

        self.update_ui()  # เรียกอัปเดต UI

    def start(self):
        self.running = True
        self.thread = Thread(target=self.long_running_task)
        self.thread.start()

    def stop(self):
        self.running = False

    def long_running_task(self):
        while self.running:
            self.queue.put("Running...")
            time.sleep(1)  # รอ 1 วินาที
        self.queue.put("Stopped")

    def update_ui(self):
        try:
            while True:
                message = self.queue.get_nowait()
                self.label.config(text=message)  # อัปเดต UI
        except Empty:
            pass
        self.root.after(100, self.update_ui)  # เรียก update_ui ทุก 100 มิลลิวินาที

root = tk.Tk()
app = App(root)
root.mainloop()
```

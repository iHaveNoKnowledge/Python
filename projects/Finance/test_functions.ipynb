# Progress Bar

```python
import tkinter as tk
from tkinter import ttk

def start_progress():
    progress['value'] = 0
    max_value = 100
    step = 1
    
    for i in range(max_value):
        progress['value'] += step
        root.update_idletasks()  # อัปเดต UI เพื่อให้เห็นความคืบหน้า
        root.after(50)  # หน่วงเวลา 50 มิลลิวินาที
        
def increase_progress():
    progress['value'] += 10
    root.update_idletasks()
    percentage.set(progress['value'])
        

root = tk.Tk()
root.title("Progressbar Example")
percentage = tk.StringVar(value=0)

percentage_label = tk.Label(root, textvariable=percentage)
percentage_label.pack()

progress = ttk.Progressbar(root, orient=tk.HORIZONTAL, length=300, mode='determinate', maximum=100)
progress.pack(pady=20)

inc_btn = tk.Button(root, text="Increase Progress", command=increase_progress)
inc_btn.pack(pady=15)

start_button = tk.Button(root, text="Start Progress", command=start_progress)
start_button.pack(pady=10)

root.mainloop()
```

# Threading and interupt

### 01example

```python
import threading
import time

# สร้าง Event สำหรับส่งสัญญาณการหยุด Thread
stop_event = threading.Event()

def worker():
    while not stop_event.is_set():
        print("Thread is running...")
        time.sleep(1)
    print("Thread is stopping...")

# สร้าง Thread
thread = threading.Thread(target=worker)

# เริ่มการทำงานของ Thread
thread.start()

# ให้ Thread ทำงานเป็นเวลา 5 วินาที
time.sleep(5)

# ส่งสัญญาณให้หยุดการทำงานของ Thread
stop_event.set()

# รอให้ Thread หยุดทำงาน
thread.join()

print("Main program finished.")
```

### 02 My example

```python
import threading
import time

# สร้าง Event สำหรับส่งสัญญาณการหยุด Thread
stop_event = threading.Event()

total_task = 10
current_task = 0


def worker():
    global current_task
    global percentage_task
    while not stop_event.is_set():
        current_task += 1
        percentage_task = (current_task / total_task) 
        if not percentage_task > 1: 
            print("Thread is running...")
            print(f"Current task: {percentage_task * 100}%")
        time.sleep(1)
        
    print("Thread is stopping...")

# สร้าง Thread
thread = threading.Thread(target=worker)

# เริ่มการทำงานของ Thread
thread.start()
print(threading.enumerate())
print(thread.daemon)
#* ให้ Thread ทำงานจนกว่างานจะจบ
while current_task <= total_task:
    time.sleep(1)

# ส่งสัญญาณให้หยุดการทำงานของ Thread
stop_event.set()

# รอให้ Thread หยุดทำงาน
thread.join()

print("Main program finished.")
print(threading.enumerate())
print(thread.ident)
```

### logger
1. TRACE: ใช้สำหรับข้อมูลการดีบักที่ละเอียดมาก
2. DEBUG: ใช้สำหรับข้อมูลการดีบักที่สำคัญ
3. INFO: ใช้สำหรับข้อมูลทั่วไปที่ต้องการแสดง
4. SUCCESS: ใช้สำหรับการรายงานสถานะสำเร็จ
5. WARNING: ใช้สำหรับการเตือนภัย
6. ERROR: ใช้สำหรับข้อผิดพลาด
7. CRITICAL: ใช้สำหรับข้อผิดพลาดที่ร้ายแรงมาก

โดยปกติแล้วจะตั้งค่าการล็อกจากระดับที่ต่ำไปสูง เช่น หากกำหนดระดับที่เป็น WARNING ข้อมูลที่มีระดับต่ำกว่านี้ (เช่น INFO, DEBUG, TRACE) จะไม่ถูกบันทึก.

```python
import threading
import time
from loguru import logger

def worker(stop_event, total_task):
    current_task = 0
    
    while not stop_event.is_set():
        current_task += 1
        percentage_task = (current_task / total_task) * 100
        
        if percentage_task <= 100: 
            print("Thread is running...")
            print(f"Current task: {percentage_task:.2f}%")
        else:
            break
            
        time.sleep(1)
    
    print("Thread is stopping...")

#* Create logger 
logger.add("test_functions_jupyter.log", format="{time} {level} {message}", level="INFO")

# Initialize stop_event
stop_event = threading.Event()

# Total number of tasks
total_task = 10

# Create and start the thread
thread = threading.Thread(target=worker, args=(stop_event, total_task))
thread.start()

# Keep the main thread running until the worker thread finishes
thread.join()

print("Main program finished.")
print(threading.enumerate())
print(thread.ident)
```

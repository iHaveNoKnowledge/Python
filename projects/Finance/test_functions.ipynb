# Progress Bar

```python
import tkinter as tk
from tkinter import ttk

def start_progress():
    progress['value'] = 0
    max_value = 100
    step = 1
    
    for i in range(max_value):
        progress['value'] += step
        root.update_idletasks()  # อัปเดต UI เพื่อให้เห็นความคืบหน้า
        root.after(50)  # หน่วงเวลา 50 มิลลิวินาที
        
def increase_progress():
    progress['value'] += 10
    root.update_idletasks()
    percentage.set(progress['value'])
        

root = tk.Tk()
root.title("Progressbar Example")
percentage = tk.StringVar(value=0)

percentage_label = tk.Label(root, textvariable=percentage)
percentage_label.pack()

progress = ttk.Progressbar(root, orient=tk.HORIZONTAL, length=300, mode='determinate', maximum=100)
progress.pack(pady=20)

inc_btn = tk.Button(root, text="Increase Progress", command=increase_progress)
inc_btn.pack(pady=15)

start_button = tk.Button(root, text="Start Progress", command=start_progress)
start_button.pack(pady=10)

root.mainloop()
```

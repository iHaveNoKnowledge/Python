# Progress Bar

```python
import tkinter as tk
from tkinter import ttk

def start_progress():
    progress['value'] = 0
    max_value = 100
    step = 1
    
    for i in range(max_value):
        progress['value'] += step
        root.update_idletasks()  # อัปเดต UI เพื่อให้เห็นความคืบหน้า
        root.after(50)  # หน่วงเวลา 50 มิลลิวินาที
        
def increase_progress():
    progress['value'] += 10
    root.update_idletasks()
    percentage.set(progress['value'])
        

root = tk.Tk()
root.title("Progressbar Example")
percentage = tk.StringVar(value=0)

percentage_label = tk.Label(root, textvariable=percentage)
percentage_label.pack()

progress = ttk.Progressbar(root, orient=tk.HORIZONTAL, length=300, mode='determinate', maximum=100)
progress.pack(pady=20)

inc_btn = tk.Button(root, text="Increase Progress", command=increase_progress)
inc_btn.pack(pady=15)

start_button = tk.Button(root, text="Start Progress", command=start_progress)
start_button.pack(pady=10)

root.mainloop()
```

### 01example

```python
import threading
import time

# สร้าง Event สำหรับส่งสัญญาณการหยุด Thread
stop_event = threading.Event()

def worker():
    while not stop_event.is_set():
        print("Thread is running...")
        time.sleep(1)
    print("Thread is stopping...")

# สร้าง Thread
thread = threading.Thread(target=worker)

# เริ่มการทำงานของ Thread
thread.start()

# ให้ Thread ทำงานเป็นเวลา 5 วินาที
time.sleep(5)

# ส่งสัญญาณให้หยุดการทำงานของ Thread
stop_event.set()

# รอให้ Thread หยุดทำงาน
thread.join()

print("Main program finished.")
```

### 02 My example

```python
import threading
import time

# สร้าง Event สำหรับส่งสัญญาณการหยุด Thread
stop_event = threading.Event()

total_task = 10
current_task = 0


def worker():
    global current_task
    global percentage_task
    while not stop_event.is_set():
        current_task += 1
        percentage_task = (current_task / total_task) 
        if not percentage_task > 1: 
            print("Thread is running...")
            print(f"Current task: {percentage_task * 100}%")
        time.sleep(1)
        
    print("Thread is stopping...")

# สร้าง Thread
thread = threading.Thread(target=worker)

# เริ่มการทำงานของ Thread
thread.start()
print(threading.enumerate())
print(thread.daemon)
#* ให้ Thread ทำงานจนกว่างานจะจบ
while current_task <= total_task:
    time.sleep(1)

# ส่งสัญญาณให้หยุดการทำงานของ Thread
stop_event.set()

# รอให้ Thread หยุดทำงาน
thread.join()

print("Main program finished.")
print(threading.enumerate())
print(thread.ident)
```

### logger
1. TRACE: ใช้สำหรับข้อมูลการดีบักที่ละเอียดมาก
2. DEBUG: ใช้สำหรับข้อมูลการดีบักที่สำคัญ
3. INFO: ใช้สำหรับข้อมูลทั่วไปที่ต้องการแสดง
4. SUCCESS: ใช้สำหรับการรายงานสถานะสำเร็จ
5. WARNING: ใช้สำหรับการเตือนภัย
6. ERROR: ใช้สำหรับข้อผิดพลาด
7. CRITICAL: ใช้สำหรับข้อผิดพลาดที่ร้ายแรงมาก

โดยปกติแล้วจะตั้งค่าการล็อกจากระดับที่ต่ำไปสูง เช่น หากกำหนดระดับที่เป็น WARNING ข้อมูลที่มีระดับต่ำกว่านี้ (เช่น INFO, DEBUG, TRACE) จะไม่ถูกบันทึก.

```python
import threading
import time
from loguru import logger

def worker(stop_event, total_task):
    current_task = 0
    
    while not stop_event.is_set():
        current_task += 1
        percentage_task = (current_task / total_task) * 100
        
        if percentage_task <= 100: 
            print("Thread is running...")
            print(f"Current task: {percentage_task:.2f}%")
        else:
            break
            
        time.sleep(1)
    
    print("Thread is stopping...")

#* Create logger 
logger.add("test_functions_jupyter.log", format="{time} {level} {message}", level="INFO")

# Initialize stop_event
stop_event = threading.Event()

# Total number of tasks
total_task = 10

# Create and start the thread
thread = threading.Thread(target=worker, args=(stop_event, total_task))
thread.start()

# Keep the main thread running until the worker thread finishes
thread.join()

print("Main program finished.")
print(threading.enumerate())
print(thread.ident)
```

# queue

```python
import tkinter as tk
import threading
import queue
import time

# ฟังก์ชันที่ทำงานใน thread แยก
def worker(q):
    for i in range(5):
        time.sleep(1)  # จำลองการทำงานที่ใช้เวลา
        q.put(f"Message {i+1}")  # ส่งข้อความเข้า queue

# ฟังก์ชันที่ดึงข้อมูลจาก queue และอัพเดท UI
def process_queue():
    try:
        while True:
            message = q.get_nowait()  # ดึงข้อมูลจาก queue
            log_textbox.insert(tk.END, message + "\n")  # แสดงข้อความใน Text widget
    except queue.Empty:
        pass
    
    # เรียก process_queue ใหม่ทุก 100 มิลลิวินาที
    root.after(100, process_queue)

# สร้าง root window สำหรับ Tkinter
root = tk.Tk()
root.title("Queue Example")

# สร้าง Text widget สำหรับแสดง log
log_textbox = tk.Text(root, width=40, height=10)
log_textbox.pack(pady=20)

# สร้าง queue
q = queue.Queue()

# เริ่ม thread แยก
threading.Thread(target=worker, args=(q,), daemon=True).start()

# เริ่มการประมวลผล queue
root.after(100, process_queue)

# เริ่ม loop ของ Tkinter
root.mainloop()
```

```python
import tkinter as tk
from threading import Thread
import queue
import time

def long_running_task(task_queue):
    time.sleep(5)  # Simulate long-running task
    task_queue.put("Task Completed")

def process_queue(task_queue):
    try:
        while True:
            message = task_queue.get_nowait()
            label.config(text=message)
    except queue.Empty:
        root.after(100, process_queue, task_queue)

root = tk.Tk()
label = tk.Label(root, text="Starting...")
label.pack()

task_queue = queue.Queue()
thread = Thread(target=long_running_task, args=(task_queue,))
thread.start()

root.after(100, process_queue, task_queue)
root.mainloop()
```

### Print by win32

```python
import win32api
import win32print
import win32ui
import win32con

def print_pdf(pdf_path):
    printer_name = win32print.GetDefaultPrinter() #* เลือกprinter
    handle_Printer = win32print.OpenPrinter(printer_name) #* เชื่อมต่อกับ printer return handler
    try:
        #* สร้างเอกสารการพิมพ์ ด้วย Device Context (เกี่ยวกับการแสดงผลหน้าจอ device)
        hPrinterDC = win32ui.CreateDC() #* สร้าง obj สำหรับจัดการ Device Context
        hPrinterDC.CreatePrinterDC(printer_name) #* สร้าง Device Context ลงบน printer ที่เชื่อม
        hPrinterDC.StartDoc("Autoinv: Printing inv") #* ส่งข้อความไปโชว์ที่ Context ของ printer หรือคือหน้าจอ printer นั่นเอง
        hPrinterDC.StartPage() #* เริ่มหน้าใหม่ในการพิมเอกสาร
        
        #* ใช้ win32api เพื่อพิมพ์ไฟล์ PDF
        win32api.ShellExecute(0, "print", pdf_path, None, ".", win32con.SW_HIDE)
        
        hPrinterDC.EndPage()
        hPrinterDC.EndDoc()
    finally:
        win32print.ClosePrinter(handle_Printer)

print_pdf("output.pdf")
```

# Device Context

```python
import win32ui
import win32con
import win32api
import win32gui

def draw_on_screen():
    # Get the device context (DC) for the entire screen
    hdc = win32gui.GetDC(0)
    
    # Create a compatible DC for drawing
    compatible_dc = win32ui.CreateDCFromHandle(hdc)
    
    # Set the background and text colors
    compatible_dc.SetBkColor(win32api.RGB(255, 255, 255))  # White background
    compatible_dc.SetTextColor(win32api.RGB(0, 0, 0))      # Black text
    
    # Draw the text at the specified position
    compatible_dc.TextOut(100, 100, "Hello, World!")
    
    # Release the DC when done
    win32gui.ReleaseDC(0, hdc)
    
    # Delete the compatible DC to free resources
    compatible_dc.DeleteDC()

draw_on_screen()
```
